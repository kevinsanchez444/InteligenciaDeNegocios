```python
# eliminando la columna redundante
df_plataforma = pd.read_csv('estudiantes_cursos.csv', index_col=0)
df_plataforma.head()
```

```python
# limpieza de datos duplicados
df_plataforma.drop_duplicates(inplace=True)

# Muestra el DataFrame 
display(df_plataforma)
```

```python
# filtrado condicional basico
df_plataforma[df_plataforma["Lecciones_Completadas"] > 40]
```

```python
# filtrado por coincidencia multiple (isin)
df_plataforma[df_plataforma["Curso"].isin(["Python para datos", "Arquitectura Cloud"])]
```

```python
# filtrado por rango de desempeño (between)
df_plataforma[df_plataforma["Puntaje_Evaluacion"].between(70.0,85.0)]
```

```python
#consulta legibles con .query
df_plataforma.query("Modalidad == 'Asincrónico' and Horas_Estudio > 25")
```

```python
# filtrado compuesto con busqueda de texto con el operador &
df_plataforma[(df_plataforma["Nivel_Educativo"] == "Profesional") & (df_plataforma["Curso"].str.startswith("B"))]
```

```python
#imputacion de datos faltantes con estrategias mixtas
df_plataforma.fillna({"Horas_Estudio": df_plataforma["Horas_Estudio"].mean(), "Puntaje_Evaluacion": 0.0}, inplace=True)
```

```python
# transformacion vectorizada con .apply
def evaluar_estado(nota):
    if nota >= 70.0:
        return "Aprobado"
    else:
        return "Reprobado"


df_plataforma["Estado_Materia"] = df_plataforma["Puntaje_Evaluacion"].apply(evaluar_estado)


df_plataforma[["Puntaje_Evaluacion", "Estado_Materia"]].head()
```

```python
# 1) Renombrar solo si existen
renombres = {
    "Modalidad": "Formato_Clase",
    "Certificado_Generado": "Graduado"
}
for col_orig, col_nuevo in renombres.items():
    if col_orig in df_plataforma.columns:
        df_plataforma = df_plataforma.rename(columns={col_orig: col_nuevo})

# 2) Reordenar con seguridad
columnas_ordenadas = [
    "Estado_Materia", "ID_Estudiante", "Curso", "Nivel_Educativo",
    "Horas_Estudio", "Lecciones_Completadas", "Puntaje_Evaluacion",
    "Formato_Clase", "Graduado"
]

for c in columnas_ordenadas:
    if c not in df_plataforma.columns:
        df_plataforma[c] = None

df_plataforma = df_plataforma[columnas_ordenadas]
print(df_plataforma.head())

```

```python
# agrupacion y estadisticas globales por curso
resultado_agrupado = df_plataforma.groupby('Curso').agg({
    'Horas_Estudio': 'mean',
    'Lecciones_Completadas': 'sum'
})

# Muestra el resultado de la agrupación
resultado_agrupado
```

```python
# calculo de dispersion con funcion personalizada
def rango(series):
    return series.max() - series.min()

# 2. Agrupar por Nivel_Educativo y aplicar la función personalizada a Puntaje_Evaluacion
resultado_rango = df_plataforma.groupby('Nivel_Educativo')['Puntaje_Evaluacion'].agg(rango)

# 3. Mostrar el resultado
resultado_rango

```

```python
# Agrupamiento multinivel por Curso y Formato_Clase, calculando la suma de horas y promedio de puntaje
resultado_multinivel = df_plataforma.groupby(['Curso', 'Formato_Clase']).agg({
    'Horas_Estudio': 'sum',
    'Puntaje_Evaluacion': 'mean'
})

# Muestra el resultado de la agrupación multinivel
resultado_multinivel
```
